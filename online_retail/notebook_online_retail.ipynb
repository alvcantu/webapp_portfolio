```python
def categorize_quantity(quantity):
    if quantity > 0:
        return "Greater than 0"
    elif quantity == 0:
        return "0"
    else:
        return "Less than 0"

# Assuming df is your DataFrame
df['PriceCategory'] = df['UnitPrice'].apply(categorize_quantity)

# Count the occurrences of each category
category_counts = df['PriceCategory'].value_counts()

# Display the counts
print(category_counts)
```

```python
import re

def categorize_stock_code(code):
    if re.match(r'^\d{5}$', code):
        return "5 digit integer"
    elif re.match(r'^\d{5}[A-Za-z]{2}$', code):
        return "5 digit integer + 2 letters"
    elif re.match(r'^DCGS', code):
        return "Starts with DCGS"
    elif re.match(r'^gift', code, re.IGNORECASE):
        return "Starts with gift"
    elif re.match(r'^\d{5}[A-Za-z]$', code):
        return "5 digit integer + 1 letter"
    else:
        return "Other"

# Assuming df is your DataFrame
df['StockCodeCategory'] = df['StockCode'].apply(categorize_stock_code)
```

```python
# Assuming df is your DataFrame with the 'StockCodeCategory' column already created

# Count the occurrences of each category
category_counts = df['StockCodeCategory'].value_counts()

# Display the counts
print(category_counts)
```
