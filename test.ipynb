```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(df['subscribed_y'], df['predicted_subscribed_y'])
print(f'ROC-AUC Score: {roc_auc}')
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# Convert probabilities to 0/1 based on a threshold
threshold = 0.5
predicted_binary = (df['predicted_subscribed_y'] >= threshold).astype(int)

precision = precision_score(df['subscribed_y'], predicted_binary)
recall = recall_score(df['subscribed_y'], predicted_binary)
f1 = f1_score(df['subscribed_y'], predicted_binary)

print(f'Precision: {precision}, Recall: {recall}, F1 Score: {f1}')
```

```python
from sklearn.metrics import log_loss

log_loss_value = log_loss(df['subscribed_y'], df['predicted_subscribed_y'])
print(f'Log Loss: {log_loss_value}')
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(df['subscribed_y'], predicted_binary)
print(f'Confusion Matrix:\n {cm}')
```

```python
from sklearn.metrics import brier_score_loss

brier_score = brier_score_loss(df['subscribed_y'], df['predicted_subscribed_y'])
print(f'Brier Score: {brier_score}')
```
