```python
# SQL Query to fetch Total_Sales and Predicted_Sales4
sql_query = '''
SELECT TransactionID, UnitPrice, Quantity, UnitPrice*Quantity as Total_Sales, Predicted_Sales4 FROM ONR_FactTransactions;
'''

cursor.execute(sql_query)
data = cursor.fetchall()

# Convert to DataFrame for easier manipulation
df = pd.DataFrame(data, columns=['TransactionID', 'UnitPrice', 'Quantity', 'Total_Sales', 'Predicted_Sales4'])

# Close the cursor and connection
cursor.close()
mydb.close()

print(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size
plt.figure(figsize=(10, 6))

# Plot histograms for both columns
sns.histplot(df['Predicted_Sales4'], label='Predicted Sales4', kde=True, color='blue', bins=50, stat="density")
sns.histplot(df['Total_Sales'], label='Total Sales', kde=True, color='orange', bins=50, stat="density")

# Set labels and title
plt.xlabel('Sales Value')
plt.ylabel('Density')
plt.title('Comparison of Predicted Sales4 and Total Sales Distributions')

# Add legend
plt.legend()

# Display the plot
plt.show()
```
